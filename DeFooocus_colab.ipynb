```python
#@title DeFooocus — Install & Launch
import os
import sys
import subprocess

def run(*args):
    subprocess.run(list(args), check=True)

print("[DeFooocus] Preparing Python 3.11...")
run(sys.executable, "-m", "pip", "install", "-U", "uv")

uv = "/usr/local/bin/uv"
run(uv, "python", "install", "3.11")

env = "/content/defooocus-py311"
if not os.path.exists(env + "/bin/python"):
    run(uv, "venv", "--python", "3.11", "--seed", env)

repo = "/content/DeFooocus"
if not os.path.exists(repo):
    run("git", "clone",
        "https://github.com/ehristoforu/DeFooocus.git", repo)

os.chdir(repo)
python = env + "/bin/python"

print("[DeFooocus] Installing requirements...")
run(python, "-m", "pip", "install",
    "-r", "requirements_versions.txt")

print("[DeFooocus] Starting...")
run(python, "entry_with_update.py",
    "--share", "--theme", "dark", "--attention-split")
```
